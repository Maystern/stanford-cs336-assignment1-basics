## Problem (unicode2): Unicode Encodings (3 points)

(a) $$ \textbf{compression\_rate} = \frac{\textbf{len}(\text{encoded\_str})}{\textbf{len}(\text{str})} $$ satisfies $$ \text{utf-8} < \text{utf-16} < \text{utf-32} $$ We don't want the compression rate is very large, because the context of encoded string may be difficult to solve for LLM.

````python
def test_encode_info(test_str, encoding):
    encoded_str = test_str.encode(encoding)
    print("#####" + encoding + "#####")
    print("test_string = " + test_str)
    print("result\t", encoded_str)
    print("list\t", list(encoded_str))
    print("len\t", str(len(list(encoded_str))))
    print("")

test_sting = "hello! こんにちは!"
print("test_string = " + test_sting)
print("len(test_string) = " + str(len(test_sting)))
print("")

test_encode_info(test_sting, "utf-8")
test_encode_info(test_sting, "utf-16")
test_encode_info(test_sting, "utf-32")
````

(b) 

```python
bytestring = "你".encode("utf-8") # b'\xe4\xbd\xa0'
```
because utf-8 unicode bytes include ASCLL and other CHARACTERS, not all characters which can encoded by utf-8 contain 1 byte each.

````python
def decode_utf8_bytes_to_str_wrong(bytestring: bytes):
    return "".join([bytes([b]).decode("utf-8") for b in bytestring])

# utf-8 编码的构成，可能是一个字节、两个字节或者是多个字节
# "你" 字符串由 [11100100] [10111101] [10100000] 三个字节构成
# 其中 11100100 中的 3 个 1 表示后续的 3 个字节构成这个 char，而单独的 [11100100] 不可能对应一个 1 字节的 char

bytestring = "你".encode("utf-8")
try:
    print(f"string = \"{bytestring}\"")
    decode_utf8_bytes_to_str_wrong(bytestring)
except Exception as e:
    print(e)
````

````python
# 在 unicode 编码中，存在代理码点，0xD800-0xDFFF 不对应任何字符
bytestring = b'\xd8\x00'
try:
    print(f"string = \"{bytestring}\"")
    bytestring.decode("utf-8")
except Exception as e:
    print(e)
````
